```python
import pandas as pd

df = pd.read_csv("flipkart_com-ecommerce_sample.csv")

df.head()
```

```python
df = df[['product_name',
         'product_category_tree',
         'brand',
         'retail_price',
         'discounted_price',
         'product_rating']]
```

```python
df.head()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df['product_rating'].unique()
```

```python
(df['product_rating'] == 'No rating available').sum()
```

```python
df['product_rating'] = df['product_rating'].replace(
    'No rating available',
    None
)
```

```python
df['product_rating'] = pd.to_numeric(
    df['product_rating'],
    errors='coerce'
)
```

```python
df.info()
```

```python
df['discount_percentage'] = (
    (df['retail_price'] - df['discounted_price'])
    / df['retail_price']
) * 100
```

```python
df['discount_percentage'] = df['discount_percentage'].round(2)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
df['retail_price'].hist(bins=50)
plt.title('Distribution of Retail Prices')
plt.xlabel('Retail Price')
plt.ylabel('Number of Products')
plt.show()
```

```python
df.groupby('brand')['retail_price'] \
.mean() \
.sort_values(ascending=False) \
.head(10) \
.plot(kind='bar', figsize=(10,5))

plt.title('Top Brands by Average Price')
plt.ylabel('Average Price')
plt.show()
```

```python
top_brands = df['brand'].value_counts().head(10)

plt.figure(figsize=(12,6))

top_brands.sort_values().plot(kind='barh')

plt.title('Top 10 Brands by Product Count', fontsize=16)
plt.xlabel('Number of Products')
plt.ylabel('Brand')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
df['discount_percentage'].hist(bins=30)

plt.title('Discount Percentage Distribution')
plt.xlabel('Discount Percentage')
plt.ylabel('Number of Products')

plt.show()
```

```python
sample_df = df.sample(3000)

plt.figure(figsize=(12,6))

plt.scatter(
    sample_df['retail_price'],
    sample_df['discount_percentage'],
    alpha=0.5
)

plt.title('Price vs Discount Analysis')
plt.xlabel('Retail Price (₹)')
plt.ylabel('Discount Percentage')

plt.show()
```
